Assignment 1: Application program two different lists

Declare two different lists:

```
color = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]
```

Now for each item in color print out each item in fruits with the respective color, so the output looks similar to:

```
red apple
red banana
red cherry
big apple
big banana
big cherry
tasty apple
tasty banana
tasty cherry
```
Tip: Nested loops!

````python
print("\n" + "=" * 50)
print("ASSIGNMENT 1: TWO DIFFERENT LISTS WITH NESTED LOOPS")
print("=" * 50)

color = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

print("Colors:", color)
print("Fruits:", fruits)
print("\nCombining colors and fruits using nested loops:")

for c in color:
    for f in fruits:
        print(f"{c} {f}")

print("\nWhat we learned:")
print("- Nested loops allow us to iterate over multiple lists")
print("- The outer loop runs once for each color")
print("- The inner loop runs completely for each iteration of the outer loop")
````

### Assignment 2: Application program 2-dim arrays (list) of numbers

Lists with sub-lists are actually the Python version of 2-dim arrays.
Declare a list with four sub-lists:

```
T = [[11, 12, 5, 2], [15, 6, 10], [10, 8, 12, 5], [12,15,8,6]]
print(T)
```

Try:

```
T[1][2] = 99
```

Which element is changed to the value 99? Print out and see if you were right?
Add a nested for-loop so for each sub-list, slist, in T print each element, elem, in sub-list.


````python
print("\n" + "=" * 50)
print("ASSIGNMENT 2: 2-DIMENSIONAL ARRAYS (LISTS)")
print("=" * 50)

T = [[11, 12, 5, 2], [15, 6, 10], [10, 8, 12, 5], [12, 15, 8, 6]]
print("Original 2D list:")
print(T)

print(f"\nBefore change: T[1][2] = {T[1][2]}")
T[1][2] = 99
print(f"After T[1][2] = 99: T[1][2] = {T[1][2]}")

print("\nThe changed element is in:")
print("- Row 1 (second row, 0-indexed)")
print("- Column 2 (third column, 0-indexed)")
print("- So the element '10' in [15, 6, 10] was changed to 99")

print("\nPrinting 2D list using nested loops:")
for row_index, slist in enumerate(T):
    print(f"Row {row_index}:", end=" ")
    for elem in slist:
        print(elem, end=" ")
    print()  # New line after each row

print("\nAlternative method with indices:")
for i in range(len(T)):
    for j in range(len(T[i])):
        print(f"T[{i}][{j}] = {T[i][j]}")
````

### Assignment 3: Random numbers

You are to do a program, which can initialize an integer list holding 10 elements of random numbers in the interval `[0,100]` using the module random.

````python
import random
print("\n" + "=" * 50)
print("ASSIGNMENT 3: RANDOM NUMBERS")
print("=" * 50)

print("Creating a list of 10 random numbers between 0 and 100:")

data = []  # create an empty list
while len(data) < 10:
    randNo = random.randint(0, 100)
    data.append(randNo)

print("Random numbers list:", data)
print(f"Length of list: {len(data)}")

print("\nWhat we learned:")
print("- random.randint(a, b) generates integers from a to b (inclusive)")
print("- We can build lists dynamically using append()")
print("- while loops continue until a condition is met")
````

Assignment 4: Random function

Create your own random method, which returns a list of random numbers.

```
def rand_func():
. . . . . . .
```



````python
# Assignment 4: Random function
print("\n" + "=" * 50)
print("ASSIGNMENT 4: CUSTOM RANDOM FUNCTION")
print("=" * 50)

def rand_func():
    """
    Creates and returns a list of 10 random numbers between 0 and 100
    """
    random_list = []
    while len(random_list) < 10:
        randNo = random.randint(0, 100)
        random_list.append(randNo)
    return random_list

# Test the function
random_result = rand_func()
print("Result from rand_func():", random_result)

print("\nWhat we learned:")
print("- Functions can create and return data structures")
print("- We can encapsulate repetitive code in functions")
````

````python
print("\n" + "=" * 50)
print("EXTRA ASSIGNMENT: ENHANCED RANDOM FUNCTION")
print("=" * 50)

def enhanced_rand_func(size=10, max_value=100):
    """
    Creates and returns a list of random numbers
    Parameters:
    size: number of elements in the list (default 10)
    max_value: maximum value for random numbers (default 100)
    """
    random_list = []
    while len(random_list) < size:
        randNo = random.randint(0, max_value)
        random_list.append(randNo)
    return random_list

# Test the enhanced function
print("Default parameters (10 numbers, max 100):")
result1 = enhanced_rand_func()
print(result1)

print("\nCustom parameters (5 numbers, max 50):")
result2 = enhanced_rand_func(5, 50)
print(result2)

print("\nCustom parameters (15 numbers, max 10):")
result3 = enhanced_rand_func(15, 10)
print(result3)
````

Here we want to calculate some numbers called Fibonacci numbers and save them in a list. The numbers are defined as:

```
i y[i]
0 0
1 1
2 1 = 1+0
3 2 = 1+1
4 3 = 1+2
5 5 = 3+2
6 8 = 5+3
7 13= 8+5
```

As you can see a Fibonacci number from a certain point is the sum of the two previous numbers. Do a loop so the respective elements in y is assigned the sum of the two previous elements.!!

````python
print("\n" + "=" * 50)
print("ASSIGNMENT X: FIBONACCI SEQUENCE")
print("=" * 50)

def fibonacci_sequence(n=8):
    """
    Generate Fibonacci sequence with n numbers
    F(0) = 0, F(1) = 1, F(n) = F(n-1) + F(n-2) for n > 1
    """
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]

    # Initialize the sequence with first two numbers
    fib = [0, 1]

    # Generate remaining numbers
    for i in range(2, n):
        next_fib = fib[i-1] + fib[i-2]
        fib.append(next_fib)

    return fib

# Generate and display Fibonacci sequence
print("Fibonacci sequence (first 8 numbers):")
y = fibonacci_sequence(8)

print("i  | y[i] | Calculation")
print("---|------|------------------")
for i in range(len(y)):
    if i == 0:
        calc = "0 (base case)"
    elif i == 1:
        calc = "1 (base case)"
    else:
        calc = f"{y[i-1]} + {y[i-2]} = {y[i]}"
    print(f"{i:2} | {y[i]:4} | {calc}")

print(f"\nComplete Fibonacci sequence: {y}")

# Extended Fibonacci sequence
print("\nExtended Fibonacci sequence (first 15 numbers):")
extended_fib = fibonacci_sequence(15)
print(extended_fib)

# Alternative implementation using a loop
print("\nAlternative implementation using while loop:")
def fibonacci_while_loop(n):
    """Alternative Fibonacci implementation using while loop"""
    y = []
    i = 0
    while i < n:
        if i == 0:
            y.append(0)
        elif i == 1:
            y.append(1)
        else:
            y.append(y[i-1] + y[i-2])
        i += 1
    return y

fib_while = fibonacci_while_loop(10)
print("Fibonacci using while loop:", fib_while)
````
